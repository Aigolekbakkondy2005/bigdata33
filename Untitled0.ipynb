```python
import numpy as np
data = np.array([
    5000, 7000, 8000, 6000, 9000,
    7500, 8500, 6500, 10000, 500000
])
mean = np.mean(data)
std = np.std(data)
z_scores = (data - mean) / std
anomalies = data[np.abs(z_scores) > 2]
print("Орташа мән:", mean)
print("Z-score:", z_scores)
print("Аномалиялар:", anomalies)
```

```python
anomalies = data[np.abs(z_scores) > 2]
print("Anyqtalgan anomalya:", anomalies)
```

```python
indices = np.where(np.abs(z_scores) > 2)[0]
print("Anomalya indeksteri:", indices)
print("Anomalya manderi:", data[indices])
```

```python
clean_data = data[np.abs(z_scores) <= 2]
new_mean = np.mean(clean_data)
print("Bastapqy ortasha man:", np.mean(data))
print("Anomalyasyz ortasha man:", new_mean)
```

```python
import numpy as np
np.random.seed(52)
n = 1_000_000
data = np.column_stack([
    np.random.randint(1, 100001, n),
    np.random.randint(1, 10001, n),
    np.random.uniform(10, 100000, n),
    np.random.randint(1, 11, n),
    np.random.uniform(0, 30, n),
    np.random.randint(1, 11, n)
])
print("Massiv olshemi:", data.shape)
print("Massiv olshemi", data.ndim)
print("algashqy 5 jazba:")
print(data[:5])
```

```python
unique_data = np.unique(data, axis=0)
print("Bastapqy jazbalar:", data.shape[0])
print("Biregei jazbalar:", unique_data.shape[0])
print("Qaitalanatyn jazbalar:", data.shape[0] - unique_data.shape[0])
```

```python
data_with_nan = data.copy()
missing_index = np.random.choice(n, 1000, replace=False)
data_with_nan[missing_index, 2] = np.nan
print("Jetispeitin mander sany:",
      np.sum(np.isnan(data_with_nan)))
```

```python
data_with_nan[0, 2] = -500
data_with_nan[1, 3] = -2
data_with_nan[2, 4] = 150
data_with_nan[3, 5] = 99
```

```python
invalid_amount = data_with_nan[:, 2] < 0
invalid_quantity = data_with_nan[:, 3] < 1
invalid_discount = (data_with_nan[:, 4] < 0) | (data_with_nan[:, 4] > 100)
invalid_region = (data_with_nan[:, 5] < 1) | (data_with_nan[:, 5] > 10)

print("Qate satyp alu:", np.sum(invalid_amount))
print("Qate tauar sany:", np.sum(invalid_quantity))
print("Qate jenildek:", np.sum(invalid_discount))
print("Qate aimaq:", np.sum(invalid_region))
```

```python
median_amount = np.nanmedian(data_with_nan[:, 2])
data_with_nan[np.isnan(data_with_nan[:, 2]), 2] = median_amount
print("Nan sandar:", np.sum(np.isnan(data_with_nan)))
```

```python
clean_data = np.unique(data_with_nan)
print("Tazalangan dataset:", clean_data.shape)
```

```python
filtered_data = data_with_nan[data_with_nan[:, 2] > 50000]
print(filtered_data)
```

```python
total_income = np.sum(data_with_nan[:, 2])

print("Jalpy tabys:", total_income)
```

```python
client_ids = data_with_nan[:, 0].astype(int)
amounts = data_with_nan[:, 2]
unique_clients = np.unique(client_ids)
client_average = []
for client in unique_clients:
    client_amounts = amounts[client_ids == client]
    client_average.append(np.mean(client_amounts))
client_average = np.array(client_average)
print("Klientterdin ortasha styp alu somasy:",
      np.mean(client_average))
```

```python
regions = data_with_nan[:, 5].astype(int)

for region in np.unique(regions):
    region_sales = np.sum(data_with_nan[regions == region, 2])
    print("Aimaq", region, "satylym:", region_sales)
```

```python
amount = data_with_nan[:, 2]
print("Ortasha man:", np.mean(amount))
print("Mediana:", np.median(amount))
print("Min:", np.min(amount))
print("Max:", np.max(amount))
print("Auytqu:", np.std(amount))
print("25%:", np.percentile(amount, 25))
print("50%:", np.percentile(amount, 50))
print("75%:", np.percentile(amount, 75))
```

```python
Q1 = np.percentile(amount, 25)
Q3 = np.percentile(amount, 75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
anomalies = data_with_nan[
    (data_with_nan[:, 2] < lower) |
    (data_with_nan[:, 2] > upper)
]
print("Anomalialar:", anomalies.shape[0])
```

```python
region_sales = {}
for region in np.unique(regions):
    region_sales[region] = np.sum(
        data_with_nan[regions == region, 2]
    )
best_region = max(region_sales, key=region_sales.get)
print("Satylym kolemi jogary:", best_region)
print("Satylym:", region_sales[best_region])
```
